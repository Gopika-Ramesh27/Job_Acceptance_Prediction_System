```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_HR_job_placement_dataset.csv")
```

```python
df.shape
```

```python
def get_experience_cat(exp):
    if exp == 0:
        return 'Fresher'
    elif exp <= 2:
        return 'Junior'
    else:
        return 'Senior'

df['experience_category'] = df['years_of_experience'].apply(get_experience_cat)
```

```python
df['experience_category'].value_counts
```

```python
df['academic_band'] = pd.cut(
    df['degree_percentage'],
    bins=[0, 65, 80, 100],
    labels=['Low', 'Average', 'High']
)
```

```python
df['skills_match_level'] = pd.cut(
    df['skills_match_percentage'],
    bins=[0, 60, 80, 100],
    labels=['Low', 'Medium', 'High']
)
```

```python
df['avg_interview_score'] = (df['technical_score'] + df['aptitude_score'] + df['communication_score']) / 3

df['interview_performance_category'] = pd.cut(
    df['avg_interview_score'],
    bins=[0, 50, 75, 100],
    labels=['Needs Improvement', 'Good', 'Excellent']
)
```

```python
df['placement_probability_score'] = (
    0.4 * (df['skills_match_percentage'] / 100) +
    0.3 * (df['degree_percentage'] / 100) +
    0.3 * (df['avg_interview_score'] / 100)
).round(2)
```

```python
df[['years_of_experience', 'experience_category', 
    'degree_percentage', 'academic_band', 
    'skills_match_percentage', 'skills_match_level', 
    'interview_performance_category', 'placement_probability_score']].head()
```
