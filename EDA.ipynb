```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("cleaned_HR_job_placement_dataset.csv")
```

```python
df.shape
```

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='status', y='technical_score', palette='Set2')
plt.title('Technical Interview Score vs Job Acceptance')
plt.xlabel('Placement Status')
plt.ylabel('Technical Score')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.kdeplot(data=df, x='skills_match_percentage', hue='status', common_norm=False, fill=True, palette='coolwarm')
plt.title('Skills Match Percentage Distribution by Status')
plt.xlabel('Skills Match %')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='company_tier', hue='status', palette='viridis')
plt.title('Acceptance Rate by Company Tier')
plt.xlabel('Company Tier')
plt.ylabel('Candidate Count')
plt.legend(title='Status')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='years_of_experience', hue='status', palette='Blues')
plt.title('Years of Experience vs Placement Success')
plt.xlabel('Years of Experience')
plt.ylabel('Candidate Count')
plt.legend(title='Status')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
numeric_df = df.select_dtypes(include=['number'])
sns.heatmap(numeric_df.corr(), cmap='Blues', annot=False)
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```
