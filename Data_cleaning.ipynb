```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\DELL\Downloads\HR_Job_Placement_Dataset.csv")
```

```python
df.duplicated().sum()
```

```python
df.duplicated()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
for col in df.select_dtypes(include=['number']).columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
for col in df.select_dtypes(include=['object','string']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum().sum()
```

```python
df.dtypes
```

```python
df['notice_period_days'].value_counts
```

```python
df['notice_period_days'] = df['notice_period_days'].astype(int)
```

```python
df['employment_gap_months'].value_counts
```

```python
df['employment_gap_months'] = df['employment_gap_months'].astype(int)
```

```python
for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].astype(str).str.strip().str.title()
```
